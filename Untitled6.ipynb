```python
import unicodedata

NAME_ABBREVIATIONS = {
    "pvt": "private",
    "pvtltd": "private limited",
    "ltd": "limited",
    "lim": "limited",
    "corp": "corporation",
    "corpor": "corporation",
    "inc": "incorporated",
    "incorp": "incorporated",
    "co": "company",
    "cos": "companies",
    "llc": "limited liability company",
    "llp": "limited liability partnership",
    "lp": "limited partnership",
    "plc": "public limited company",
    "ent": "enterprise",
    "ents": "enterprises",
    "enterp": "enterprise",
    "ind": "industries",
    "inds": "industries",
    "indl": "industrial",
    "mfg": "manufacturing",
    "mfr": "manufacturer",
    "mfrs": "manufacturers",
    "manuf": "manufacturing",
    "trdg": "trading",
    "trd": "trading",
    "trdr": "trader",
    "trdrs": "traders",
    "svc": "service",
    "svcs": "services",
    "serv": "services",
    "tech": "technology",
    "techs": "technologies",
    "technol": "technology",
    "engg": "engineering",
    "engr": "engineering",
    "equip": "equipment",
    "eqpt": "equipment",
    "elec": "electrical",
    "elect": "electrical",
    "intl": "international",
    "intnl": "international",
    "natl": "national",
    "assoc": "association",
    "assn": "association",
    "asso": "association",
    "grp": "group",
    "ctr": "center",
    "cntr": "center",
    "dept": "department",
    "dist": "distribution",
    "distr": "distribution",
    "whsl": "wholesale",
    "whsle": "wholesale",
    "ret": "retail",
    "prod": "products",
    "prods": "products",
    "soln": "solution",
    "solns": "solutions",
    "sys": "systems",
    "mgmt": "management",
    "consult": "consulting",
    "comm": "communications",
    "dev": "development",
    "const": "construction",
    "constr": "construction",
    "trans": "transport",
    "transp": "transport",
    "transpt": "transport",
    "log": "logistics",
    "pharm": "pharmaceutical",
    "pharma": "pharmaceutical",
    "med": "medical",
    "hosp": "hospital",
    "lab": "laboratory",
    "labs": "laboratories",
    "inst": "institute",
    "instt": "institute",
    "univ": "university",
}
GENERAL_ADDRESS_ABBREVIATIONS = {
    "rd": "road",
    "rds": "roads",
    "st": "street",
    "sts": "streets",
    "ave": "avenue",
    "av": "avenue",
    "blvd": "boulevard",
    "ln": "lane",
    "dr": "drive",
    "ct": "court",
    "cir": "circle",
    "trl": "trail",
    "ter": "terrace",
    "pl": "place",
    "pkwy": "parkway",
    "hwy": "highway",
    "expy": "expressway",
    "expwy": "expressway",
    "fwy": "freeway",
    "jct": "junction",
    "xing": "crossing",
    "sq": "square",
    "bldg": "building",
    "bld": "building",
    "fl": "floor",
    "flr": "floor",
    "apt": "apartment",
    "apts": "apartments",
    "ste": "suite",
    "rm": "room",
    "nr": "near",
    "opp": "opposite",
    "adj": "adjacent",
    "no": "number",
    "num": "number",
    "nos": "numbers",
    "ext": "extension",
    "extn": "extension",
    "ctr": "center",
    "cntr": "center",
    "mkt": "market",
    "cmplx": "complex",
}
INDIA_ADDRESS_ABBREVIATIONS = {
    "dist": "district",
    "distt": "district",
    "vill": "village",
    "vlg": "village",
    "po": "post office",
    "ps": "police station",
    "tq": "taluka",
    "tal": "taluka",
    "sec": "sector",
    "ph": "phase",
    "indl": "industrial",
    "hno": "house number",
}
US_ADDRESS_ABBREVIATIONS = {
    "aly": "alley",
    "n": "north",
    "s": "south",
    "e": "east",
    "w": "west",
    "ne": "northeast",
    "nw": "northwest",
    "se": "southeast",
    "sw": "southwest",
}
FRANCE_ADDRESS_ABBREVIATIONS = {
    "av": "avenue",
    "bd": "boulevard",
    "rte": "route",
}
LEGAL_SUFFIX_PHRASES = [
    ["private", "limited"],
    ["public", "limited", "company"],
    ["limited", "liability", "company"],
    ["limited", "liability", "partnership"],
    ["limited", "partnership"],
    ["corporation"],
    ["incorporated"],
    ["company"],
    ["limited"],
    ["llc"],
    ["llp"],
    ["plc"],
    ["lp"],
]

def clean_input(text):
    if text is None:
        return ""
    try:
        if pd.isna(text):
            return ""
    except Exception:
        pass
    return str(text)


def unicode_normalize(text):
    text = clean_input(text)
    text = unicodedata.normalize("NFKC", text)
    return text.lower().strip()

def accent_fold(text):
    text = unicodedata.normalize("NFKD", text)
    return "".join(ch for ch in text if not unicodedata.combining(ch))


def replace_tokens(text, mapping):
    output = []
    for token in text.split():
        replacement = mapping.get(token, token)
        output.extend(replacement.split())
    return " ".join(output)

def collapse_spaces(text):
    return re.sub(r"\s+", " ", text).strip()

def normalize_business_name(text):

    text = unicode_normalize(text)
    text = re.sub(r"^\s*m\s*[/.\-]?\s*s[.\s]*", "", text)
    text = text.replace("&", " and ")
    text = text.replace("+", " and ")
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)
    text = text.replace("_", " ")
    text = collapse_spaces(text)
    text = replace_tokens(text, NAME_ABBREVIATIONS)
    return collapse_spaces(text)

def remove_suffix_tokens(tokens, suffix):
    n = len(suffix)
    if len(tokens) >= n and tokens[-n:] == suffix:
        return tokens[:-n], True
    return tokens, False


def create_core_name(normalized_name):
    tokens = normalized_name.split()
    changed = True
    while changed and tokens:
        changed = False
        for suffix in LEGAL_SUFFIX_PHRASES:
            tokens, removed = remove_suffix_tokens(tokens, suffix)
            if removed:
                changed = True
                break
    return " ".join(tokens)

def normalize_address(text, country=None):
    text = unicode_normalize(text)
    text = text.replace("&", " and ")
    text = re.sub(r"[/,.;:#()\-]+", " ", text)

    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)
    text = text.replace("_", " ")

    text = collapse_spaces(text)
    mapping = GENERAL_ADDRESS_ABBREVIATIONS.copy()
    country_norm = (str(country).strip().lower() if country is not None else "")
    if country_norm in {"india","in","ind"}:
        mapping.update(INDIA_ADDRESS_ABBREVIATIONS)

    elif country_norm in {"us","usa","united states","united states of america"}:
        mapping.update(US_ADDRESS_ABBREVIATIONS)

    elif country_norm in {"france","fr"}:
        mapping.update(FRANCE_ADDRESS_ABBREVIATIONS)

    text = replace_tokens(text, mapping)

    return collapse_spaces(text)

def extract_numbers(text):
    text = clean_input(text)
    return re.findall(r"\d+", text)


def normalize_record(business_name,business_address,country):
    name_norm = normalize_business_name(business_name)
    name_core = create_core_name(name_norm)

    address_norm = normalize_address(business_address,country)

    return {"name_norm": name_norm,"name_core": name_core,"name_ascii": accent_fold(name_norm),"address_norm": address_norm,"address_ascii": accent_fold(address_norm),"numeric_tokens": extract_numbers(business_address),"combined_norm":collapse_spaces(name_core + " " + address_norm)}

def add_normalized_columns(df):

    df = df.copy()
    df["name_norm"] = (df["business_name"].fillna("").apply(normalize_business_name))
    df["name_core"] = (df["name_norm"].apply(create_core_name))
    df["name_ascii"] = (df["name_norm"].apply(accent_fold))
    df["address_norm"] = [normalize_address(address, country) for address, country in zip(df["business_address"].fillna(""),df["country"].fillna(""))]
    df["address_ascii"] = (df["address_norm"].apply(accent_fold))
    df["numeric_tokens"] = (df["business_address"].fillna("").apply(extract_numbers))
    df["combined_norm"] = (df["name_core"] + " " + df["address_norm"]).str.strip()

    return df
df1=add_normalized_columns(df1)
df2=add_normalized_columns(df2)
df3=add_normalized_columns(df3)
```

```python
def get_matching_pattern(match_ids):
    if not match_ids.strip():
        return "No Match"

    ids = [x.strip() for x in match_ids.split(",") if x.strip()]
    has_s2 = any(x.startswith("S2-") for x in ids)
    has_s3 = any(x.startswith("S3-") for x in ids)
    if has_s2 and has_s3:
        return "Both S2 and S3"
    elif has_s2:
        return "S2 Only"
    elif has_s3:
        return "S3 Only"
    else:
        return "Unknown"

gt["matching_pattern"] = (gt["matched_entity_ids"].apply(get_matching_pattern))
pattern_summary = (gt["matching_pattern"].value_counts().reset_index())
pattern_summary.columns = ["Matching_Pattern","Count"]
pattern_summary["Percentage"] = (pattern_summary["Count"] / len(gt) * 100)
display(pattern_summary)
fig = px.bar(pattern_summary,x="Matching_Pattern",y="Count",text="Count",title="S1 Matching Pattern: S2 vs S3")
fig.show()
```

```python
SOURCE_FILE = "df_1.csv"
CANDIDATE_FILES = [("df_2.csv", "df2"),("df_3.csv", "df3")]
OUTPUT_FILE = "cheap_candidates.csv"
POSITIVE_FILE = "positive_pairs.csv"
SOURCE_CHUNK_SIZE = 20000
MAX_BUCKET_SIZE = 20
MAX_CANDIDATES_PER_SOURCE = 12
INPUT_COLUMNS = ["entity_id","country_norm","name_core","name_ascii","address_ascii","numeric_tokens"]
RULES = [
    {"name": "exact_core_name","keys": ["country_norm","name_core_key"],"weight": 5},
    {"name": "exact_ascii_name","keys": ["country_norm","name_ascii_key"],"weight": 4},
    {"name": "exact_address","keys": ["country_norm","address_exact_key"],"weight": 5},
    {"name": "name6_number","keys": ["country_norm","name6","first_number"],"weight": 4},
    {"name": "name4_address10","keys": ["country_norm","name4","address10"],"weight": 3},
    {"name": "name4_numeric","keys": ["country_norm","name4","numeric_key"],"weight": 4}
]
```

```python
def prepare_keys(df):
    df = df.copy()
    for column in INPUT_COLUMNS:
        df[column] = (df[column].fillna("").astype(str).str.strip())

    name_core_clean = (df["name_core"].str.replace(" ", "", regex=False))
    name_ascii_clean = (df["name_ascii"].str.replace(" ", "", regex=False))
    address_clean = (df["address_ascii"].str.replace(" ", "", regex=False))
    df["name_core_key"] = df["name_core"]
    df["name_ascii_key"] = df["name_ascii"]
    df["address_exact_key"] = df["address_ascii"]
    df["name4"] = name_core_clean.str[:4]
    df["name6"] = name_core_clean.str[:6]
    df["address10"] = address_clean.str[:10]
    df["numeric_key"] = (df["numeric_tokens"].str.replace(r"\s+", "", regex=True))
    df["first_number"] = (df["numeric_tokens"].str.extract(r"(\d+)",expand=False).fillna(""))
    return df


def non_empty_mask(df, columns):
    mask = pd.Series(True,index=df.index)
    for column in columns:
        mask &= df[column].ne("")

    return mask


def prepare_candidate_file(file_name):
    print("\nLoading {}".format(file_name))
    df = pd.read_csv(file_name,dtype=str,keep_default_na=False,usecols=INPUT_COLUMNS)
    df = df.drop_duplicates(subset=["entity_id"])
    df = prepare_keys(df)
    for index, rule in enumerate(RULES):
        bucket_column = "_bucket_{}".format(index)
        valid = non_empty_mask(df,rule["keys"])
        df[bucket_column] = (MAX_BUCKET_SIZE + 1)
        if valid.any():
            counts = (df.loc[valid].groupby(rule["keys"],sort=False)["entity_id"].transform("size"))
            df.loc[valid,bucket_column] = counts.astype("int32")

        usable = int((df[bucket_column]<= MAX_BUCKET_SIZE).sum())
        print("{}: {:,} usable candidate rows".format(rule["name"],usable))

    needed_columns = {"entity_id"}
    for rule in RULES:
        needed_columns.update(rule["keys"])

    for index in range(len(RULES)):
        needed_columns.add("_bucket_{}".format(index))

    return df[list(needed_columns)].copy()


def get_pairs_for_rule( source,candidate,candidate_source,rule,rule_index):
    keys = rule["keys"]
    left_valid = non_empty_mask(source,keys)
    bucket_column = "_bucket_{}".format(rule_index)
    right_valid = (candidate[bucket_column]<= MAX_BUCKET_SIZE)
    right_valid &= non_empty_mask(candidate,keys)
    left = source.loc[left_valid,["entity_id"] + keys].rename(columns={"entity_id":"source1_entity_id"})
    right = candidate.loc[right_valid,["entity_id"] + keys].rename(columns={"entity_id":"candidate_entity_id"})
    if left.empty or right.empty:
        return None

    pairs = left.merge(right,on=keys,how="inner",sort=False)
    if pairs.empty:
        return None

    pairs = pairs[["source1_entity_id","candidate_entity_id"]]
    pairs["candidate_source"] = (candidate_source)
    pairs["block_score"] = (rule["weight"])
    pairs["method"] = (rule["name"])
    return pairs

def select_best_candidates(parts):
    if not parts:
        return pd.DataFrame(columns=["source1_entity_id","candidate_entity_id","candidate_source"])
    
    pairs = pd.concat(parts,ignore_index=True)
    pairs = (pairs.groupby(["source1_entity_id","candidate_entity_id","candidate_source"],as_index=False).agg(block_score=("block_score","sum"),methods_found=("method","nunique")))
    pairs = pairs.sort_values(by=["source1_entity_id","block_score","methods_found","candidate_entity_id"],ascending=[True,False,False,True])
    pairs = (pairs.groupby("source1_entity_id",sort=False).head(MAX_CANDIDATES_PER_SOURCE))
    return pairs[["source1_entity_id","candidate_entity_id","candidate_source"]]
```

```python
print("Preparing candidate indexes")
candidate_data = []
for file_name, source_name in CANDIDATE_FILES:
    candidate_data.append((prepare_candidate_file(file_name),source_name))

positive_set = None
found_positives = set()
if os.path.exists(POSITIVE_FILE):
    positive_pairs = pd.read_csv(POSITIVE_FILE,dtype=str,keep_default_na=False,usecols=["source1_entity_id","candidate_entity_id"]).drop_duplicates()
    positive_set = set(positive_pairs.itertuples(index=False,name=None))
    print("\nKnown positive pairs: {:,}".format(len(positive_set)))

if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

header_written = False
processed_sources = 0
written_pairs = 0
print("\nStarting source blocking")
for source_chunk in pd.read_csv(SOURCE_FILE,dtype=str,keep_default_na=False,usecols=INPUT_COLUMNS,chunksize=SOURCE_CHUNK_SIZE):
    source_chunk = (source_chunk.drop_duplicates(subset=["entity_id"]))
    source_chunk = prepare_keys(source_chunk)
    source_needed = {"entity_id"}
    for rule in RULES:
        source_needed.update(rule["keys"])

    source_chunk = source_chunk[list(source_needed)]
    parts = []
    for rule_index, rule in enumerate(RULES):
        for candidate, candidate_source in (candidate_data):
            result = get_pairs_for_rule(source_chunk,candidate,candidate_source,rule,rule_index)
            if result is not None:
                parts.append(result)

    selected = select_best_candidates(parts)
    if not selected.empty:
        selected.to_csv(OUTPUT_FILE,mode="a",header=not header_written,index=False)
        header_written = True
        written_pairs += len(selected)
        if positive_set is not None:
            for pair in selected[["source1_entity_id","candidate_entity_id"]].itertuples(index=False,name=None):
                if pair in positive_set:
                    found_positives.add(pair)

    processed_sources += len(source_chunk)
    print("Processed {:,} source rows | ""Total candidate pairs {:,}".format(processed_sources,written_pairs))

if not header_written:
    pd.DataFrame(columns=["source1_entity_id","candidate_entity_id","candidate_source"]).to_csv(OUTPUT_FILE,index=False)

print("\nBlocking finished")
print("Source entities processed: {:,}".format(processed_sources))
print("Candidate pairs written: {:,}".format(written_pairs))
print("Maximum candidates per source:",MAX_CANDIDATES_PER_SOURCE)
print("Output:",OUTPUT_FILE)
if positive_set is not None:
    retained = len(found_positives)
    total = len(positive_set)
    recall = (retained / total if total > 0 else 0.0)
    print("\nPositive pairs retained: " "{:,} / {:,}".format(retained,total))
    print("Blocking recall: {:.2%}".format(recall))
```

```python
PAIR_FILE = "cheap_candidates.csv"
FEATURE_FILE = "candidate_pairs_training_features.csv"
DF1_FILE = "df_1.csv"
DF2_FILE = "df_2.csv"
DF3_FILE = "df_3.csv"
POSITIVE_FILE = "positive_pairs.csv"
GT_FILE = "train_ground_truth.tsv"
CHUNK_SIZE = 100000
BATCH_SIZE = 2048
MAX_EPOCHS = 8
VAL_MOD = 20
VAL_BUCKET = 0
EARLY_STOPPING_PATIENCE = 2
MIN_IMPROVEMENT = 0.0001
RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
IDS = ["source1_entity_id", "candidate_entity_id"]
ENTITY_COLUMNS = ["entity_id","country_norm","name_core","name_ascii","address_ascii","numeric_tokens","combined_norm"]
FEATURE_COLUMNS = ["rule_exact_core_name","rule_exact_ascii_name","rule_exact_address","rule_name6_number","rule_name4_address10","rule_name4_numeric","methods_found","block_score","numeric_exact","combined_exact","name_length_similarity","address_length_similarity"]

print("Loading entity files...")
df1 = pd.read_csv(DF1_FILE,dtype=str,keep_default_na=False,usecols=ENTITY_COLUMNS).drop_duplicates("entity_id")
df2 = pd.read_csv(DF2_FILE,dtype=str,keep_default_na=False,usecols=ENTITY_COLUMNS).drop_duplicates("entity_id")
df3 = pd.read_csv(DF3_FILE,dtype=str,keep_default_na=False,usecols=ENTITY_COLUMNS).drop_duplicates("entity_id")
source_lookup = df1.rename(columns={"entity_id": "source1_entity_id","country_norm": "country_norm_s","name_core": "name_core_s","name_ascii": "name_ascii_s","address_ascii": "address_ascii_s","numeric_tokens": "numeric_tokens_s","combined_norm": "combined_norm_s"})
candidate2_lookup = df2.rename(columns={"entity_id": "candidate_entity_id","country_norm": "country_norm_c","name_core": "name_core_c","name_ascii": "name_ascii_c","address_ascii": "address_ascii_c","numeric_tokens": "numeric_tokens_c","combined_norm": "combined_norm_c"})
candidate3_lookup = df3.rename(columns={"entity_id": "candidate_entity_id","country_norm": "country_norm_c","name_core": "name_core_c","name_ascii": "name_ascii_c","address_ascii": "address_ascii_c","numeric_tokens": "numeric_tokens_c","combined_norm": "combined_norm_c"})
positive_pairs = pd.read_csv(POSITIVE_FILE,dtype=str,keep_default_na=False,usecols=IDS).drop_duplicates()
positive_pairs["label"] = np.float32(1.0)
print("Known positive pairs: {:,}".format(len(positive_pairs)))
def length_similarity(left, right):
    left_length = left.str.len().to_numpy(dtype=np.float32)
    right_length = right.str.len().to_numpy(dtype=np.float32)
    minimum = np.minimum(left_length, right_length)
    maximum = np.maximum(left_length, right_length)
    result = np.zeros(len(left), dtype=np.float32)
    valid = maximum > 0
    result[valid] = minimum[valid] / maximum[valid]
    return result

def build_features(pairs):
    pairs = pairs.copy()
    pairs["_row_order"] = np.arange(len(pairs))
    pairs = pairs.merge(source_lookup,on="source1_entity_id",how="left",validate="many_to_one")
    df2_pairs = pairs[pairs["candidate_source"] == "df2"].merge(candidate2_lookup,on="candidate_entity_id",how="left",validate="many_to_one")
    df3_pairs = pairs[pairs["candidate_source"] == "df3"].merge(candidate3_lookup,on="candidate_entity_id",how="left",validate="many_to_one")
    unknown = pairs[~pairs["candidate_source"].isin(["df2", "df3"])]
    if not unknown.empty:
        bad_sources = unknown["candidate_source"].drop_duplicates().tolist()
        raise ValueError("Unknown candidate_source values: {}".format(bad_sources))
    data = pd.concat([df2_pairs, df3_pairs],ignore_index=True).sort_values("_row_order").reset_index(drop=True)
    text_columns = ["country_norm_s","country_norm_c","name_core_s","name_core_c","name_ascii_s","name_ascii_c","address_ascii_s","address_ascii_c","numeric_tokens_s","numeric_tokens_c","combined_norm_s","combined_norm_c"]
    for column in text_columns:
        data[column] = data[column].fillna("").astype(str).str.strip()
    name_core_s = data["name_core_s"]
    name_core_c = data["name_core_c"]
    name_ascii_s = data["name_ascii_s"]
    name_ascii_c = data["name_ascii_c"]
    address_ascii_s = data["address_ascii_s"]
    address_ascii_c = data["address_ascii_c"]
    numeric_tokens_s = data["numeric_tokens_s"]
    numeric_tokens_c = data["numeric_tokens_c"]
    name_core_clean_s = name_core_s.str.replace(" ", "", regex=False)
    name_core_clean_c = name_core_c.str.replace(" ", "", regex=False)
    address_clean_s = address_ascii_s.str.replace(" ", "", regex=False)
    address_clean_c = address_ascii_c.str.replace(" ", "", regex=False)
    numeric_key_s = numeric_tokens_s.str.replace(r"\s+", "", regex=True)
    numeric_key_c = numeric_tokens_c.str.replace(r"\s+", "", regex=True)
    first_number_s = (numeric_tokens_s.str.extract(r"(\d+)", expand=False).fillna(""))
    first_number_c = (numeric_tokens_c.str.extract(r"(\d+)", expand=False).fillna(""))
    country_same = (data["country_norm_s"].ne("") & data["country_norm_s"].eq(data["country_norm_c"]))
    name4_s = name_core_clean_s.str[:4]
    name4_c = name_core_clean_c.str[:4]
    name6_s = name_core_clean_s.str[:6]
    name6_c = name_core_clean_c.str[:6]
    address10_s = address_clean_s.str[:10]
    address10_c = address_clean_c.str[:10]
    rule_exact_core_name = (country_same & name_core_s.ne("") & name_core_s.eq(name_core_c))
    rule_exact_ascii_name = (country_same & name_ascii_s.ne("") & name_ascii_s.eq(name_ascii_c))
    rule_exact_address = (country_same & address_ascii_s.ne("") & address_ascii_s.eq(address_ascii_c))
    rule_name6_number = (country_same & name6_s.ne("") & name6_s.eq(name6_c) & first_number_s.ne("") & first_number_s.eq(first_number_c))
    rule_name4_address10 = (country_same & name4_s.ne("") & name4_s.eq(name4_c) & address10_s.ne("") & address10_s.eq(address10_c))
    rule_name4_numeric = (country_same & name4_s.ne("") & name4_s.eq(name4_c) & numeric_key_s.ne("") & numeric_key_s.eq(numeric_key_c))

    data["rule_exact_core_name"] = rule_exact_core_name.astype(np.float32)
    data["rule_exact_ascii_name"] = rule_exact_ascii_name.astype(np.float32)
    data["rule_exact_address"] = rule_exact_address.astype(np.float32)
    data["rule_name6_number"] = rule_name6_number.astype(np.float32)
    data["rule_name4_address10"] = rule_name4_address10.astype(np.float32)
    data["rule_name4_numeric"] = rule_name4_numeric.astype(np.float32)
    data["methods_found"] = (data["rule_exact_core_name"] + data["rule_exact_ascii_name"] + data["rule_exact_address"] + data["rule_name6_number"] + data["rule_name4_address10"] + data["rule_name4_numeric"])
    data["block_score"] = (5.0 * data["rule_exact_core_name"] + 4.0 * data["rule_exact_ascii_name"] + 5.0 * data["rule_exact_address"] + 4.0 * data["rule_name6_number"] + 3.0 * data["rule_name4_address10"] + 4.0 * data["rule_name4_numeric"]).astype(np.float32)
    data["numeric_exact"] = (numeric_tokens_s.ne("") & numeric_tokens_s.eq(numeric_tokens_c)).astype(np.float32)
    data["combined_exact"] = (data["combined_norm_s"].ne("") & data["combined_norm_s"].eq(data["combined_norm_c"])).astype(np.float32)
    data["name_length_similarity"] = length_similarity(name_core_clean_s,name_core_clean_c)
    data["address_length_similarity"] = length_similarity(address_clean_s,address_clean_c)
    feature_data = data[["source1_entity_id","candidate_entity_id","candidate_source"] + FEATURE_COLUMNS]
    feature_data = feature_data.merge(positive_pairs,on=IDS,how="left",validate="many_to_one")
    feature_data["label"] = (feature_data["label"].fillna(0).astype(np.float32))
    return feature_data

required_pair_columns = {"source1_entity_id","candidate_entity_id","candidate_source"}
pair_sample = pd.read_csv(PAIR_FILE,nrows=10,dtype=str,keep_default_na=False)
missing = required_pair_columns - set(pair_sample.columns)
if missing:
    raise ValueError("Missing columns in {}: {}".format(PAIR_FILE,", ".join(sorted(missing))))
bad_sources = set(pair_sample["candidate_source"]) - {"df2", "df3"}
if bad_sources:
    raise ValueError("Unexpected candidate_source values in {}: {}".format(PAIR_FILE,sorted(bad_sources)))
if os.path.exists(FEATURE_FILE):
    os.remove(FEATURE_FILE)

print("\nBuilding Blocking4-compatible features...")
first_chunk = True
total_feature_rows = 0
total_positive_rows = 0
for pair_chunk in pd.read_csv(PAIR_FILE,chunksize=CHUNK_SIZE,dtype=str,keep_default_na=False):
    feature_chunk = build_features(pair_chunk)
    feature_chunk.to_csv(FEATURE_FILE,mode="a",header=first_chunk,index=False)
    first_chunk = False
    total_feature_rows += len(feature_chunk)
    total_positive_rows += int(feature_chunk["label"].sum())
    print("Prepared {:,} rows | positives {:,}".format(total_feature_rows,total_positive_rows))
print("\nFeature file created:", FEATURE_FILE)
print("Blocked candidate rows: {:,}".format(total_feature_rows))
print("Positive rows retained in candidates: {:,}".format(total_positive_rows))
del df1
del df2
del df3
del source_lookup
del candidate2_lookup
del candidate3_lookup
del positive_pairs
def get_fold(series):
    hashes = pd.util.hash_pandas_object(series,index=False).to_numpy()
    return hashes % VAL_MOD

scaler = StandardScaler()
positive_count = 0
negative_count = 0
print("\nFitting scaler...")
for chunk in pd.read_csv(FEATURE_FILE,chunksize=CHUNK_SIZE):
    fold = get_fold(chunk["source1_entity_id"])
    train = chunk[fold != VAL_BUCKET]
    if train.empty:
        continue

    X = (train[FEATURE_COLUMNS].apply(pd.to_numeric, errors="coerce").fillna(0).to_numpy(dtype=np.float32))
    scaler.partial_fit(X)
    positive_count += int(train["label"].sum())
    negative_count += int((train["label"] == 0).sum())
print("Train positives:", positive_count)
print("Train negatives:", negative_count)
print("Number of features:", len(FEATURE_COLUMNS))
print("Features:", FEATURE_COLUMNS)
if positive_count == 0:
    raise ValueError("No positive pairs survived blocking. Check blocking recall.")
class EntityMatcher(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.model = nn.Sequential(nn.Linear(input_size, 64),nn.ReLU(),nn.Dropout(0.15),nn.Linear(64, 32),nn.ReLU(),nn.Dropout(0.10),nn.Linear(32, 1))

    def forward(self, x):
        return self.model(x).squeeze(1)

model = EntityMatcher(len(FEATURE_COLUMNS)).to(device)
print("\nModel architecture:")
print(model)
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.AdamW(model.parameters(),lr=0.001,weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer,mode="max",factor=0.5,patience=1)
def train_epoch(model):
    model.train()
    total_loss = 0.0
    total_rows = 0
    for chunk in pd.read_csv(FEATURE_FILE,chunksize=CHUNK_SIZE):
        fold = get_fold(chunk["source1_entity_id"])
        chunk = chunk[fold != VAL_BUCKET]
        if chunk.empty:
            continue

        X = (chunk[FEATURE_COLUMNS].apply(pd.to_numeric, errors="coerce").fillna(0).to_numpy(dtype=np.float32))
        X = scaler.transform(X).astype(np.float32)
        y = chunk["label"].to_numpy(dtype=np.float32)
        dataset = TensorDataset(torch.from_numpy(X),torch.from_numpy(y))
        loader = DataLoader(dataset,batch_size=BATCH_SIZE,shuffle=True,pin_memory=(device.type == "cuda"))
        for xb, yb in loader:
            xb = xb.to(device, non_blocking=True)
            yb = yb.to(device, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()
            total_loss += loss.item() * len(xb)
            total_rows += len(xb)

    if total_rows == 0:
        raise ValueError("No training rows found.")
    return total_loss / total_rows

def predict_validation(model):
    model.eval()
    outputs = []
    with torch.no_grad():
        for chunk in pd.read_csv(FEATURE_FILE,chunksize=CHUNK_SIZE):
            fold = get_fold(chunk["source1_entity_id"])
            chunk = chunk[fold == VAL_BUCKET].copy()
            if chunk.empty:
                continue
            X = (chunk[FEATURE_COLUMNS].apply(pd.to_numeric, errors="coerce").fillna(0).to_numpy(dtype=np.float32))
            X = scaler.transform(X).astype(np.float32)
            probabilities = []
            for start in range(0, len(X), BATCH_SIZE):
                xb = torch.from_numpy(X[start:start + BATCH_SIZE]).to(device)
                probability = torch.sigmoid(model(xb)).cpu().numpy()
                probabilities.append(probability)
            chunk["probability"] = np.concatenate(probabilities)
            outputs.append(chunk[["source1_entity_id","candidate_entity_id","probability"]])
    if not outputs:
        raise ValueError("No validation predictions were created.")
    return pd.concat(outputs, ignore_index=True)

gt = pd.read_csv(GT_FILE,sep="\t",dtype=str,keep_default_na=False)
gt["fold"] = get_fold(gt["source1_entity_id"])
gt_val = gt[gt["fold"] == VAL_BUCKET].copy()
true_sets = {}
for _, row in gt_val.iterrows():
    matches = {value.strip() for value in row["matched_entity_ids"].split(",") if value.strip()}
    true_sets[row["source1_entity_id"]] = matches

def macro_f05(predictions, threshold):
    selected = predictions[predictions["probability"] >= threshold]
    predicted_sets = (selected.groupby("source1_entity_id")["candidate_entity_id"].apply(set).to_dict())
    scores = []
    for source_id, true_set in true_sets.items():
        predicted_set = predicted_sets.get(source_id, set())
        if len(true_set) == 0:
            scores.append(1.0 if len(predicted_set) == 0 else 0.0)
            continue

        if len(predicted_set) == 0:
            scores.append(0.0)
            continue

        true_positive = len(true_set & predicted_set)
        if true_positive == 0:
            scores.append(0.0)
            continue

        precision = true_positive / len(predicted_set)
        recall = true_positive / len(true_set)
        f05 = (1.25 * precision * recall / (0.25 * precision + recall))
        scores.append(f05)
    if not scores:
        return 0.0
    return float(np.mean(scores))

thresholds = np.arange(0.05, 0.991, 0.01)
best_score = -1.0
best_threshold = None
best_epoch = None
best_state = None
early_stopping_score = -1.0
epochs_without_improvement = 0
for epoch in range(1, MAX_EPOCHS + 1):
    loss = train_epoch(model)
    validation_predictions = predict_validation(model)
    scores = [macro_f05(validation_predictions, threshold) for threshold in thresholds]
    position = int(np.argmax(scores))
    epoch_score = float(scores[position])
    epoch_threshold = float(thresholds[position])
    scheduler.step(epoch_score)
    current_lr = optimizer.param_groups[0]["lr"]
    print("Epoch {} | loss={:.5f} | F0.5={:.6f} | threshold={:.2f} | lr={:.6f}".format(epoch,loss,epoch_score,epoch_threshold,current_lr))
    if epoch_score > best_score:
        best_score = epoch_score
        best_threshold = epoch_threshold
        best_epoch = epoch
        best_state = copy.deepcopy(model.state_dict())

    if epoch_score > early_stopping_score + MIN_IMPROVEMENT:
        early_stopping_score = epoch_score
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1
        print("No meaningful improvement: {}/{}".format(epochs_without_improvement,EARLY_STOPPING_PATIENCE))

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:
        print("\nEarly stopping triggered.")
        break

if best_state is None:
    raise ValueError("Training did not produce a valid model.")

model.load_state_dict(best_state)

print("\nBest epoch:", best_epoch)
print("Best Macro F0.5:", best_score)
print("Best threshold:", best_threshold)

torch.save({"model_state": model.state_dict(),"input_size": len(FEATURE_COLUMNS),"features": FEATURE_COLUMNS,"threshold": best_threshold,"best_f05": best_score},"entity_matcher_dl.pt")

joblib.dump(scaler,"dl_scaler.pkl")

print("\nSaved:")
print(FEATURE_FILE)
print("entity_matcher_dl.pt")
print("dl_scaler.pkl")
```

```python
SOURCE_FILE = "test_source1.csv"
CANDIDATE_FILES = [
    ("test_source2.csv", "df2"),
    ("test_source3.csv", "df3"),
]
OUTPUT_FILE = "test_candidate_features.csv"
SOURCE_CHUNK_SIZE = 20000
MAX_BUCKET_SIZE = 20
MAX_CANDIDATES_PER_SOURCE = 12
INPUT_COLUMNS = [
    "entity_id",
    "country_norm",
    "name_core",
    "name_ascii",
    "address_ascii",
    "numeric_tokens",
]
RULES = [
    {
        "name": "exact_core_name",
        "keys": ["country_norm", "name_core_key"],
        "weight": 5,
    },
    {
        "name": "exact_ascii_name",
        "keys": ["country_norm", "name_ascii_key"],
        "weight": 4,
    },
    {
        "name": "exact_address",
        "keys": ["country_norm", "address_exact_key"],
        "weight": 5,
    },
    {
        "name": "name6_number",
        "keys": ["country_norm", "name6", "first_number"],
        "weight": 4,
    },
    {
        "name": "name4_address10",
        "keys": ["country_norm", "name4", "address10"],
        "weight": 3,
    },
    {
        "name": "name4_numeric",
        "keys": ["country_norm", "name4", "numeric_key"],
        "weight": 4,
    },
]
def prepare_keys(df):
    df = df.copy()
    for column in INPUT_COLUMNS:
        df[column] = (
            df[column]
            .fillna("")
            .astype(str)
            .str.strip()
        )

    name_core_clean = df["name_core"].str.replace(" ", "", regex=False)
    address_clean = df["address_ascii"].str.replace(" ", "", regex=False)
    df["name_core_key"] = df["name_core"]
    df["name_ascii_key"] = df["name_ascii"]
    df["address_exact_key"] = df["address_ascii"]
    df["name4"] = name_core_clean.str[:4]
    df["name6"] = name_core_clean.str[:6]
    df["address10"] = address_clean.str[:10]
    df["numeric_key"] = df["numeric_tokens"].str.replace(r"\s+", "", regex=True)
    df["first_number"] = df["numeric_tokens"].str.extract(r"(\d+)", expand=False).fillna("")
    return df


def non_empty_mask(df, columns):
    mask = pd.Series(True, index=df.index)
    for column in columns:
        mask &= df[column].ne("")
    return mask

def prepare_candidate_file(file_name):
    df = pd.read_csv(
        file_name,
        dtype=str,
        keep_default_na=False,
        usecols=INPUT_COLUMNS,
    )
    df = df.drop_duplicates(subset=["entity_id"])
    df = prepare_keys(df)

    for index, rule in enumerate(RULES):
        bucket_column = f"_bucket_{index}"
        valid = non_empty_mask(df, rule["keys"])
        df[bucket_column] = MAX_BUCKET_SIZE + 1

        if valid.any():
            counts = (
                df.loc[valid]
                .groupby(rule["keys"], sort=False)["entity_id"]
                .transform("size")
            )
            df.loc[valid, bucket_column] = counts.astype("int32")

    needed_columns = {"entity_id"}
    for rule in RULES:
        needed_columns.update(rule["keys"])

    for index in range(len(RULES)):
        needed_columns.add(f"_bucket_{index}")

    return df[list(needed_columns)].copy()

def get_pairs_for_rule(source, candidate, candidate_source, rule, rule_index):
    keys = rule["keys"]
    left_valid = non_empty_mask(source, keys)
    bucket_column = f"_bucket_{rule_index}"
    right_valid = ((candidate[bucket_column] <= MAX_BUCKET_SIZE) & non_empty_mask(candidate, keys))
    left = source.loc[left_valid, ["entity_id"] + keys].rename(columns={"entity_id": "source1_entity_id"})
    right = candidate.loc[right_valid, ["entity_id"] + keys].rename(columns={"entity_id": "candidate_entity_id"})
    if left.empty or right.empty:
        return None

    pairs = left.merge(right, on=keys, how="inner", sort=False)

    if pairs.empty:
        return None

    pairs = pairs[["source1_entity_id", "candidate_entity_id",]]
    pairs["candidate_source"] = candidate_source
    pairs["block_score"] = rule["weight"]
    pairs["method"] = rule["name"]
    return pairs

def select_best_candidates(parts):
    if not parts:
        return pd.DataFrame(
            columns=[
                "source1_entity_id",
                "candidate_entity_id",
                "candidate_source",
            ]
        )

    pairs = pd.concat(parts, ignore_index=True)
    pairs = (
        pairs
        .groupby(
            [
                "source1_entity_id",
                "candidate_entity_id",
                "candidate_source",
            ],
            as_index=False,
        )
        .agg(
            block_score=("block_score", "sum"),
            methods_found=("method", "nunique"),
        )
    )

    pairs = pairs.sort_values(
        by=[
            "source1_entity_id",
            "block_score",
            "methods_found",
            "candidate_entity_id",
        ],
        ascending=[True, False, False, True],
    )

    pairs = (
        pairs
        .groupby("source1_entity_id", sort=False)
        .head(MAX_CANDIDATES_PER_SOURCE)
    )

    return pairs[[
        "source1_entity_id",
        "candidate_entity_id",
        "candidate_source",
    ]]

print("Preparing candidate indexes...")

candidate_data = [
    (prepare_candidate_file(file_name), source_name)
    for file_name, source_name in CANDIDATE_FILES
]

if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

header_written = False
processed_sources = 0
written_pairs = 0
print("Starting test-data blocking...")

for source_chunk in pd.read_csv(SOURCE_FILE, dtype=str, keep_default_na=False, usecols=INPUT_COLUMNS, chunksize=SOURCE_CHUNK_SIZE):
    source_chunk = source_chunk.drop_duplicates(subset=["entity_id"])
    source_chunk = prepare_keys(source_chunk)
    source_needed = {"entity_id"}
    for rule in RULES:
        source_needed.update(rule["keys"])

    source_chunk = source_chunk[list(source_needed)]
    parts = []

    for rule_index, rule in enumerate(RULES):
        for candidate, candidate_source in candidate_data:
            result = get_pairs_for_rule(
                source_chunk,
                candidate,
                candidate_source,
                rule,
                rule_index,
            )
            if result is not None:
                parts.append(result)

    selected = select_best_candidates(parts)
    if not selected.empty:
        selected.to_csv(
            OUTPUT_FILE,
            mode="a",
            header=not header_written,
            index=False,
        )
        header_written = True
        written_pairs += len(selected)

    processed_sources += len(source_chunk)
    print(
        f"Processed {processed_sources:,} source rows | "
        f"Candidate pairs written: {written_pairs:,}"
    )

if not header_written:
    pd.DataFrame(
        columns=[
            "source1_entity_id",
            "candidate_entity_id",
            "candidate_source",
        ]
    ).to_csv(OUTPUT_FILE, index=False)

print("\nTest-data blocking finished.")
print(f"Source entities processed: {processed_sources:,}")
print(f"Candidate pairs written: {written_pairs:,}")
print(f"Maximum candidates per source: {MAX_CANDIDATES_PER_SOURCE}")
print(f"Output: {OUTPUT_FILE}")
```

```python
import pandas as pd
import numpy as np
import gc

PAIR_FILE = "test_candidate_features.csv"
DF1_FILE = "test_norm1.csv"
DF2_FILE = "test_norm2.csv"
DF3_FILE = "test_norm3.csv"
OUTPUT_FILE = "test_model_features.csv"
CHUNK_SIZE = 200000

cols = [
    "entity_id",
    "country_norm",
    "name_core",
    "name_ascii",
    "address_ascii",
    "numeric_tokens",
    "combined_norm"
]

df1 = pd.read_csv(DF1_FILE, usecols=cols, dtype=str, keep_default_na=False)
df2 = pd.read_csv(DF2_FILE, usecols=cols, dtype=str, keep_default_na=False)
df3 = pd.read_csv(DF3_FILE, usecols=cols, dtype=str, keep_default_na=False)

for df in [df1, df2, df3]:
    for col in cols:
        df[col] = df[col].fillna("").astype(str).str.strip()

df1 = df1.rename(columns={
    "entity_id": "source1_entity_id",
    "country_norm": "s_country",
    "name_core": "s_core",
    "name_ascii": "s_ascii",
    "address_ascii": "s_address",
    "numeric_tokens": "s_numeric",
    "combined_norm": "s_combined"
})

df2["candidate_source"] = "df2"
df3["candidate_source"] = "df3"

candidate_lookup = pd.concat([df2, df3], ignore_index=True)
candidate_lookup = candidate_lookup.rename(columns={
    "entity_id": "candidate_entity_id",
    "country_norm": "c_country",
    "name_core": "c_core",
    "name_ascii": "c_ascii",
    "address_ascii": "c_address",
    "numeric_tokens": "c_numeric",
    "combined_norm": "c_combined"
})

s1_cols = [
    "source1_entity_id",
    "s_country",
    "s_core",
    "s_ascii",
    "s_address",
    "s_numeric",
    "s_combined"
]

candidate_cols = [
    "candidate_entity_id",
    "candidate_source",
    "c_country",
    "c_core",
    "c_ascii",
    "c_address",
    "c_numeric",
    "c_combined"
]

df1 = df1[s1_cols]
candidate_lookup = candidate_lookup[
    candidate_cols
]

def clean(x):
    return x.fillna("").astype(str)

def length_similarity(a, b):
    a = clean(a).str.replace(" ", "", regex=False)
    b = clean(b).str.replace(" ", "", regex=False)
    la = a.str.len().to_numpy(dtype=np.float32)
    lb = b.str.len().to_numpy(dtype=np.float32)
    minimum = np.minimum(la,lb)
    maximum = np.maximum(la,lb)
    result = np.zeros(len(a), dtype=np.float32)
    valid = maximum > 0
    result[valid] = minimum[valid] / maximum[valid]
    return result


first_chunk = True
for chunk_number, pairs in enumerate(pd.read_csv(PAIR_FILE, chunksize=CHUNK_SIZE, dtype=str, keep_default_na=False), 1):
    print("Processing chunk:", chunk_number)
    pairs = pairs.merge(df1, on="source1_entity_id", how="left")
    pairs = pairs.merge(
        candidate_lookup,
        on=[
            "candidate_entity_id",
            "candidate_source"
        ],
        how="left"
    )

    s_country = clean(pairs["s_country"])
    c_country = clean(pairs["c_country"])
    s_core = clean(pairs["s_core"])
    c_core = clean(pairs["c_core"])
    s_ascii = clean(pairs["s_ascii"])
    c_ascii = clean(pairs["c_ascii"])
    s_address = clean(pairs["s_address"])
    c_address = clean(pairs["c_address"])
    s_numeric = clean(pairs["s_numeric"])
    c_numeric = clean(pairs["c_numeric"])
    s_combined = clean(pairs["s_combined"])
    c_combined = clean(pairs["c_combined"])
    same_country = (s_country.ne("") & s_country.eq(c_country))
    s_core_clean = s_core.str.replace(" ", "",regex=False)
    c_core_clean = (c_core.str.replace(" ", "", regex=False))
    s_address_clean = s_address.str.replace(" ", "", regex=False)
    c_address_clean = c_address.str.replace(" ", "", regex=False)
    s_first_number = s_numeric.str.extract(r"(\d+)", expand=False).fillna("")
    c_first_number = c_numeric.str.extract(r"(\d+)",expand=False).fillna("")
    s_numeric_key = s_numeric.str.replace(r"\s+", "", regex=True)
    c_numeric_key = c_numeric.str.replace(r"\s+", "",regex=True)
    pairs["rule_exact_core_name"] = (same_country & s_core.ne("") &s_core.eq(c_core)).astype(np.float32)
    pairs["rule_exact_ascii_name"] = (same_country & s_ascii.ne("") & s_ascii.eq(c_ascii)).astype(np.float32)
    pairs["rule_exact_address"] = (same_country & s_address.ne("") & s_address.eq(c_address)).astype(np.float32)
    pairs["rule_name6_number"] = (same_country &s_core_clean.str[:6].ne("") & s_core_clean.str[:6].eq(c_core_clean.str[:6]) & s_first_number.ne("") & s_first_number.eq(c_first_number)).astype(np.float32)
    pairs["rule_name4_address10"] = (same_country & s_core_clean.str[:4].ne("") & s_core_clean.str[:4].eq(c_core_clean.str[:4]) & s_address_clean.str[:10].ne("") & s_address_clean.str[:10].eq( c_address_clean.str[:10])).astype(np.float32)
    pairs["rule_name4_numeric"] = (same_country & s_core_clean.str[:4].ne("") & s_core_clean.str[:4].eq( c_core_clean.str[:4]) & s_numeric_key.ne("") & s_numeric_key.eq(c_numeric_key)).astype(np.float32)

    rule_cols = [
        "rule_exact_core_name",
        "rule_exact_ascii_name",
        "rule_exact_address",
        "rule_name6_number",
        "rule_name4_address10",
        "rule_name4_numeric"
    ]

    pairs["methods_found"] = pairs[rule_cols].sum(axis=1).astype(np.float32)
    pairs["block_score"] = (5 * pairs["rule_exact_core_name"] + 4 * pairs["rule_exact_ascii_name"] + 5 * pairs["rule_exact_address"] + 4 * pairs["rule_name6_number"] + 3 * pairs["rule_name4_address10"] + 4 * pairs["rule_name4_numeric"]).astype(np.float32)
    pairs["numeric_exact"] = (s_numeric.ne("") & s_numeric.eq(c_numeric)).astype(np.float32)
    pairs["combined_exact"] = (s_combined.ne("") & s_combined.eq(c_combined)).astype(np.float32)
    pairs["name_length_similarity"] = length_similarity(s_core,c_core)
    pairs["address_length_similarity"] = length_similarity(s_address,c_address)

    features = [
        "source1_entity_id",
        "candidate_entity_id",
        "candidate_source",
        "rule_exact_core_name",
        "rule_exact_ascii_name",
        "rule_exact_address",
        "rule_name6_number",
        "rule_name4_address10",
        "rule_name4_numeric",
        "methods_found",
        "block_score",
        "numeric_exact",
        "combined_exact",
        "name_length_similarity",
        "address_length_similarity"
    ]

    output = pairs[features]
    output.to_csv(OUTPUT_FILE, mode="w" if first_chunk else "a", header=first_chunk, index=False)
    first_chunk = False
    del pairs
    del output
    gc.collect()


print("\nFeature generation completed")
print("Output:", OUTPUT_FILE)
```

```python
FEATURE_FILE = "candidate_pairs_training_features.csv"
SCALER_FILE = "dl_scaler.pkl"
GT_FILE = "train_ground_truth.tsv"

MODEL_OUTPUT = "entity_matcher_deep.pt"

CHUNK_SIZE = 100000
BATCH_SIZE = 2048

MAX_EPOCHS = 30
EARLY_STOPPING_PATIENCE = 8
MIN_IMPROVEMENT = 0.0001

VAL_MOD = 20
VAL_BUCKET = 0

LEARNING_RATE = 0.0005
WEIGHT_DECAY = 1e-4

RANDOM_STATE = 42


np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)


device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)


FEATURE_COLUMNS = ["rule_exact_core_name","rule_exact_ascii_name","rule_exact_address","rule_name6_number","rule_name4_address10","rule_name4_numeric","methods_found","block_score","numeric_exact","combined_exact","name_length_similarity","address_length_similarity"]

scaler = joblib.load(SCALER_FILE)


def get_fold(series):
    hashes = pd.util.hash_pandas_object(series,index=False).to_numpy()
    return hashes % VAL_MOD

sample = pd.read_csv(FEATURE_FILE,nrows=10)

required_columns=({"source1_entity_id","candidate_entity_id","label"}| set(FEATURE_COLUMNS))


missing = (required_columns- set(sample.columns))


if missing:
    raise ValueError("Missing columns in feature file: {}".format(sorted(missing)))


class EntityMatcherDeep(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(input_size,128),
            nn.BatchNorm1d(128),
            nn.GELU(),
            nn.Dropout(0.25),
            nn.Linear(128,64),
            nn.BatchNorm1d(64),
            nn.GELU(),
            nn.Dropout(0.20),
            nn.Linear(64,32),
            nn.BatchNorm1d(32),
            nn.GELU(),
            nn.Dropout(0.15),
            nn.Linear(32, 16),
            nn.GELU(),
            nn.Dropout(0.10),
            nn.Linear(16,1)
        )

    def forward(self, x):
        return self.model(x).squeeze(1)

model = EntityMatcherDeep(
    len(FEATURE_COLUMNS)
).to(device)

print("\nModel:")
print(model)


total_params=sum(p.numel() for p in model.parameters())

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


print("\nTotal parameters:",total_params)
print("Trainable parameters:",trainable_params)


criterion = nn.BCEWithLogitsLoss()


optimizer = torch.optim.AdamW(model.parameters(),lr=LEARNING_RATE,weight_decay=WEIGHT_DECAY)


scheduler=(torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer,mode="max",factor=0.5,patience=2))


def train_epoch():

    model.train()

    total_loss = 0.0
    total_rows = 0


    for chunk in pd.read_csv(FEATURE_FILE,chunksize=CHUNK_SIZE):
        fold = get_fold(chunk["source1_entity_id"])
        chunk = chunk[fold != VAL_BUCKET]
        if chunk.empty:
            continue
        X=(chunk[FEATURE_COLUMNS].apply(pd.to_numeric,errors="coerce").fillna(0).to_numpy(dtype=np.float32))
        X=(scaler.transform(X).astype(np.float32))

        y=(pd.to_numeric(chunk["label"],errors="coerce").fillna(0).to_numpy(dtype=np.float32))

        dataset = TensorDataset(torch.from_numpy(X),torch.from_numpy(y))


        loader = DataLoader(dataset,batch_size=BATCH_SIZE,shuffle=True,pin_memory=(device.type == "cuda"))

        for xb,yb in loader:

            xb = xb.to(device,non_blocking=True)
            yb = yb.to(device,non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = criterion(logits,yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(),5.0)
            optimizer.step()

            total_loss +=(loss.item()*len(xb))
            total_rows += len(xb)


    if total_rows == 0:
        raise ValueError("No training rows found")


    return(total_loss/total_rows)


def predict_validation():

    model.eval()

    outputs = []


    with torch.no_grad():

        for chunk in pd.read_csv(FEATURE_FILE,chunksize=CHUNK_SIZE):

            fold=get_fold(chunk["source1_entity_id"])

            chunk = chunk[fold == VAL_BUCKET].copy()
            if chunk.empty:
                continue
            X=(chunk[FEATURE_COLUMNS].apply(pd.to_numeric,errors="coerce").fillna(0).to_numpy(dtype=np.float32))

            X=(scaler.transform(X).astype(np.float32))
            probabilities = []

            for start in range(0,len(X),BATCH_SIZE):

                xb = torch.from_numpy(X[start:start + BATCH_SIZE]).to(device)
                probability = (
                    torch.sigmoid(model(xb)).cpu().numpy())
                probabilities.append(probability)
            chunk["probability"]=(np.concatenate(probabilities))
            outputs.append(chunk[["source1_entity_id","candidate_entity_id", "probability"]])


    if not outputs:
        raise ValueError("No validation predictions")
    return pd.concat(outputs,ignore_index=True)

gt = pd.read_csv(GT_FILE,sep="\t",dtype=str,keep_default_na=False)

gt["fold"] = get_fold(gt["source1_entity_id"])


gt_val=gt[gt["fold"] == VAL_BUCKET].copy()

true_sets = {}

for _, row in gt_val.iterrows():
    source_id = row["source1_entity_id"]

    true_sets[source_id] = {
        value.strip()
        for value in row["matched_entity_ids"].split(",")
        if value.strip()
    }


def macro_f05(predictions, threshold):

    selected = predictions[
        predictions["probability"] >= threshold
    ]

    predicted_sets = (
        selected
        .groupby("source1_entity_id")["candidate_entity_id"]
        .apply(set)
        .to_dict()
    )

    scores = []

    for source_id, true_set in true_sets.items():

        predicted_set = predicted_sets.get(source_id,set())

        if len(true_set) == 0:
            scores.append(1.0 if len(predicted_set) == 0 else 0.0)
            continue

        if len(predicted_set) == 0:
            scores.append(0.0)
            continue

        tp = len(true_set & predicted_set)

        if tp == 0:
            scores.append(0.0)
            continue

        precision = tp / len(predicted_set)
        recall = tp / len(true_set)

        f05 =(1.25 * precision * recall/(0.25 * precision + recall))

        scores.append(f05)

    if not scores:
        return 0.0

    return float(np.mean(scores))


thresholds = np.arange(0.30,0.991,0.01)


best_score = -1.0
best_threshold = None
best_epoch = None
best_state = None

early_stopping_score = -1.0
epochs_without_improvement = 0


print("\nStarting deep model training...")


for epoch in range(1,MAX_EPOCHS + 1):

    loss = train_epoch()


    predictions=(predict_validation())


    scores = [macro_f05(predictions, threshold)for threshold in thresholds]

    position=int(np.argmax(scores))

    epoch_score = float(scores[position])

    epoch_threshold= float(thresholds[position])
    scheduler.step(epoch_score)

    current_lr = (optimizer.param_groups[0]["lr"])
    print("Epoch {:02d}/{:02d} | ""loss={:.6f} | ""F0.5={:.6f} | ""threshold={:.2f} | ""lr={:.7f}".format(epoch,MAX_EPOCHS,loss,epoch_score,epoch_threshold,current_lr))


    if epoch_score > best_score:

        best_score = (epoch_score)

        best_threshold =( epoch_threshold)

        best_epoch = epoch

        best_state = copy.deepcopy(model.state_dict())

    if(epoch_score > early_stopping_score+ MIN_IMPROVEMENT):
        early_stopping_score=(epoch_score)
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1


        print("No improvement: {}/{}".format(  epochs_without_improvement,EARLY_STOPPING_PATIENCE))
    if(epochs_without_improvement >= EARLY_STOPPING_PATIENCE):
        print("\nEarly stopping triggered")
    break


if best_state is None:

    raise ValueError("No valid model state produced")

model.load_state_dict(best_state)


torch.save({"model_state":model.state_dict(),"input_size":len(FEATURE_COLUMNS),"features":FEATURE_COLUMNS,"threshold":best_threshold,"best_f05":best_score,"best_epoch":best_epoch,"architecture":"128-64-32-16","optimizer_state":optimizer.state_dict(),"scheduler_state":scheduler.state_dict()}, MODEL_OUTPUT)

print("\nTraining finished")

print("Best epoch:",best_epoch)

print("Best Macro F0.5:",best_score)

print("Best threshold:",best_threshold)

print("Saved:",MODEL_OUTPUT)
```
